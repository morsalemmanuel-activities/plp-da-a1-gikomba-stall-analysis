```python
# ==============================================================================
# AUTHOR: Morsal Emmanuel Alafandi
# COURSE: Advanced Future Proof With Python (PLP Academy)
# PROJECT: Wanjiku Gikomba Market Stall Analysis (Core Python Only)
# ==============================================================================

# --- DATASET SECTION ---
# 14 days of recorded daily sales revenue in Kenyan Shillings (KES)
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100, 2500, 2200, 3100, 2400, 3900, 5600, 4700]

# Mapping array replicating the 14-day calendar tracking timeline
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# Unstructured fresh-produce inventory ledger metrics
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}

# ==============================================================================
# TASK 1: BASIC SALES METRICS COMPUTATION
# ==============================================================================
total_sales = sum(daily_sales)
avg_sales = total_sales / len(daily_sales)
max_sales = max(daily_sales)
min_sales = min(daily_sales)

# Identify exact calendar index references
max_idx = daily_sales.index(max_sales)
min_idx = daily_sales.index(min_sales)

# Deduce day name and calendar week context bounds
max_day = days[max_idx]
max_week = 1 if max_idx < 7 else 2

min_day = days[min_idx]
min_week = 1 if min_idx < 7 else 2


# ==============================================================================
# TASK 2: WEEK-ON-WEEK GROWTH ANALYSIS
# ==============================================================================
def week_totals(sales):
    """
    Splits the 14-day sales data list into two individual 7-day increments.
    Returns: A tuple containing (Week 1 Total, Week 2 Total).
    """
    w1_total = sum(sales[:7])
    w2_total = sum(sales[7:])
    return (w1_total, w2_total)

# Unpack tuple values and execute percentage growth formulation
w1_total, w2_total = week_totals(daily_sales)
wow_growth = ((w2_total - w1_total) / w1_total) * 100


# ==============================================================================
# TASK 3: DAILY PERFORMANCE PROFILE CLASSIFICATION
# ==============================================================================
def classify_day(amount, average):
    """
    Classifies daily revenue relative to the overall 14-day average baseline.
    Returns 'Strong' (> +20%), 'Weak' (< -20%), or 'Normal'.
    """
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"

# Execute programmatic loops utilizing dictionary counting patterns
day_classifications = {"Strong": 0, "Normal": 0, "Weak": 0}
for amount in daily_sales:
    category = classify_day(amount, avg_sales)
    day_classifications[category] += 1


# ==============================================================================
# TASK 4: TEMPORAL PERFORMANCE SHIFTS (WEEKDAYS VS. WEEKENDS)
# ==============================================================================
weekday_sales_list = [daily_sales[i] for i in range(14) if days[i] not in ["Sat", "Sun"]]
weekend_sales_list = [daily_sales[i] for i in range(14) if days[i] in ["Sat", "Sun"]]

avg_weekday = sum(weekday_sales_list) / len(weekday_sales_list)
avg_weekend = sum(weekend_sales_list) / len(weekend_sales_list)

weekend_diff_pct = ((avg_weekend - avg_weekday) / avg_weekday) * 100


# ==============================================================================
# TASK 5: STOCK QUANTITY LEVELS AND FINANCIAL MARGINS
# ==============================================================================
def check_reorder_status(stock_dict):
    """
    Iterates over stock dictionary items to check if quantity balances
    have dropped down to or below their reorder limits.
    Returns: A list of product names requiring replenishment.
    """
    reorder_list = []
    for product, details in stock_dict.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)
    return reorder_list

low_stock_items = check_reorder_status(stock)

# Gross Margin Percentage Calculation Loop: ((Price - Cost) / Price) * 100
product_margins = {}
for product, details in stock.items():
    margin = (details["price"] - details["cost"]) / details["price"] * 100
    product_margins[product] = margin


# ==============================================================================
# TASK 6: STALL PERFORMANCE REPORT GENERATION
# ==============================================================================
report = f"""
==================================================
        WANJIKU STALL PERFORMANCE REPORT
==================================================
SALES SUMMARY OVERVIEW (14-DAY ANALYSIS)
--------------------------------------------------
Total Accumulated Revenue : KES {total_sales:,.2f}
Daily Revenue Baseline    : KES {avg_sales:,.2f}

Peak Revenue Day          : KES {max_sales:,.2f} ({max_day} | Week {max_week})
Lowest Revenue Day        : KES {min_sales:,.2f} ({min_day} | Week {min_week})
--------------------------------------------------
GROWTH & VOLATILITY PROFILE
--------------------------------------------------
Week 1 Combined Revenue  : KES {w1_total:,.2f}
Week 2 Combined Revenue  : KES {w2_total:,.2f}
Week-on-Week Growth Rate  : {wow_growth:.2f}%

Day Categories Count     : Strong: {day_classifications['Strong']} | Normal: {day_classifications['Normal']} | Weak: {day_classifications['Weak']}
--------------------------------------------------
TEMPORAL PERFORMANCE METRICS
--------------------------------------------------
Average Weekday Revenue   : KES {avg_weekday:,.2f}
Average Weekend Revenue   : KES {avg_weekend:,.2f}
Weekend vs. Weekday Shift : Weekend sales are {weekend_diff_pct:.2f}% higher
==================================================
INVENTORY & PROFIT MARGIN INSIGHTS
==================================================
Product      | Stock Level | Gross Margin %
--------------------------------------------------
"""

# Append formatted inventory ledger strings dynamically
for prod, margin in product_margins.items():
    qty = stock[prod]["qty"]
    report += f"{prod:<12} | {qty:<11} | {margin:.2f}%\n"

report += "--------------------------------------------------\n"
report += f"CRITICAL REPLENISHMENT ALERT: Please restock {', '.join(low_stock_items)} immediately.\n"
report += "=================================================="

# Print final complete formatted canvas output
print(report)
```
